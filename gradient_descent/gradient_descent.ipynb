```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
# y = 2x + 3 
# w = 2 , b = 3
```

```python
x = np.array([1,2,3,4,5])
y =np.array( 2*x + 3 ) 
```

```python
x,y
```

```python
def gradient_descent(x,y):
    w = b = 0   # weight and b (intercept)
    lr = 0.01   # learning rate
    n = 5       # no. of samples

    for i in range(10000):
        yhat =  w*x + b
        error = yhat - y
        
        dedw = (2/n)* sum(error*x)
        dedb = (2/n)*sum(error)

        w = w - lr * dedw
        b = b - lr * dedb

    mse = sum((yhat-y)**2) / n
    mae = sum(abs(yhat-y)) / n   

    return{
        "w":w,
        "b":b,
        "MSE":mse,
        "MAE":mae
    }

```

```python
def predict(xtest,ytest,w,b):
    ypredict = w*xtest + b

    return ypredict
```

```python
def MSE_prediction(ytest,ypredict):
    mse = sum((ypredict-ytest)**2)/len(ytest)
    mae = sum((ypredict-ytest))/len(ytest)
    return {
        "mse":mse,
        "mae":mae
    }
```

```python
gradient_descent(x,y)
```

```python
res = gradient_descent(x,y)
w= res["w"]
b= res["b"]
w,b   # (weights) result of model 
```

```python
xtest = np.array(np.random.randint(1,50,10))
ytest = 2*xtest + 3
ypredict = predict(xtest,ytest,w,b)
```

```python
mse_predict = MSE_prediction(ytest=ytest,ypredict=ypredict)["mse"]
mae_predict = MSE_prediction(ytest=ytest,ypredict=ypredict)["mae"]
mse_predict, mae_predict
```
